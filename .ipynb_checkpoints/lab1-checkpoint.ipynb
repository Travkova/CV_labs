```python
import cv2
import matplotlib.pyplot as plt
```

```python
img_exm = cv2.imread("image_example.jpg")
#img_rgb = cv2.cvtColor(img_exm,cv2.COLOR_BGR2RGB)

plt.imshow(img_exm)
plt.axis("off")
plt.show
```

```python
img_rgb = cv2.cvtColor(img_exm,cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show
```

```python
pixel_image = img_exm[0,0]
print(pixel_image)
```

```python
img_hsv = cv2.cvtColor(img_exm,cv2.COLOR_BGR2HSV)

plt.imshow(img_hsv)
plt.axis("off")
plt.show

print(img_hsv.shape)
```
