```python
# Импорт библиотек
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D,
    Dropout, Flatten, Dense
)

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    ConfusionMatrixDisplay
)

SEED = 42
np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)
plt.rcParams['figure.figsize'] = (10, 10)
```

### Загрузка и подготовка данных
Датасет MNIST содержит 70 000 рукописных цифр 28×28 пикселя. В датасете нет отдельной validation-выборки, поэтому формируем её самостоятельно из train с помощью стратифицированного разбиения (`stratify=y_train_all`) — это гарантирует одинаковое распределение классов во всех выборках.

**Использованные методы:**
- Загрузка датасета MNIST через `mnist.load_data()`;
- Стратифицированное разбиение (`train_test_split` с `stratify`) на train (10 000) и val (2 000) с сохранением пропорций классов;
- Нормализация пикселей: преобразуем `uint8 [0, 255]` в `float32 [0, 1]` через `.astype()` и проводим деление на 255.

```python
(x_train_all, y_train_all), (x_test, y_test) = mnist.load_data()

#10 000 изображений для обучения
all_indices = np.arange(len(y_train_all))

train_indices, remaining_indices = train_test_split(all_indices, train_size=10_000, random_state=SEED,  stratify=y_train_all)

val_indices, _ = train_test_split( remaining_indices, train_size=2_000, random_state=SEED, stratify=y_train_all[remaining_indices])

# Нормализация изображения
x_train = x_train_all[train_indices].astype(np.float32) / 255.0
y_train = y_train_all[train_indices]

x_val = x_train_all[val_indices].astype(np.float32) / 255.0
y_val = y_train_all[val_indices]

x_test = x_test.astype(np.float32) / 255.0
```

### Приведение данных к формату, требуемому каждым методом

Классические алгоритмы и свёрточные сети требуют разной структуры входа:

- **Random Forest / SVM** — двумерная матрица `(объекты, признаки)`. 
  Метод `.reshape(len(x), -1)` разворачивает изображение 28×28 в вектор из 784 пикселей.
- **CNN** — четырёхмерный тензор `(batch, height, width, channels)`. 
  Индексация `[..., np.newaxis]` добавляет ось канала (1, т.к. изображения чёрно-белые).

Атрибут `.shape` используется для контроля корректности преобразований.

```python
# Для Random Forest и SVM
X_train = x_train.reshape(len(x_train), -1)
X_test = x_test.reshape(len(x_test), -1)

# Для CNN
x_train_cnn = x_train[..., np.newaxis]
x_val_cnn = x_val[..., np.newaxis]
x_test_cnn = x_test[..., np.newaxis]

print('Обучающая:', x_train.shape)
print('Проверочная CNN:', x_val.shape)
print('Тестовая :', x_test.shape)
print('Форма данных для RF / SVM:', X_train.shape)
print('Форма данных для CNN:', x_train_cnn.shape)
```

### Визуализация изображений из обучающей выборки и подсчёт количества попавших в выборку изображений каждой цифры

```python
fig, axes = plt.subplots(2, 5, figsize=(11, 5))

for i, ax in enumerate(axes.ravel()):
    ax.imshow(x_train[i], cmap='gray')
    ax.set_title(f'Цифра: {y_train[i]}')
    ax.axis('off')

plt.suptitle('Примеры изображений MNIST')
plt.tight_layout()
plt.show()

print('Количество изображений каждой цифры:')
print(pd.Series(y_train).value_counts().sort_index().to_string())
```

### Задание 1.Обучить классификатор Random Forest на 10 000 изображениях MNIST.

В этом блоке мы обучаем ансамблевый классификатор Random Forest на 10 000 изображениях, замеряем время его обучения и оцениваем качество предсказаний на тестовой выборке.

**Используемые методы и параметры:**

* **`RandomForestClassifier(...)`** — инициализация модели леса случайных деревьев:
  * `n_estimators=120` — количество деревьев в ансамбле.
  * `random_state=SEED` — фиксация генератора случайных чисел (42) для полной воспроизводимости результатов.
  * `n_jobs=-1` — использование всех доступных ядер процессора для параллельного обучения деревьев (значительно ускоряет процесс).
* **`time.perf_counter()`** — высокоточный системный таймер. Используется для замера чистого времени обучения модели в секундах.
* **`rf.fit(X_train, y_train)`** — метод обучения модели на обучающей выборке (матрица признаков `X_train` и вектор меток `y_train`).
* **`rf.predict(X_test)`** — метод генерации предсказаний (классов цифр) для тестовой выборки.
* **`accuracy_score(y_test, rf_pred)`** — функция из библиотеки `sklearn.metrics`, вычисляющая метрику Accuracy (долю верно угаданных классов).
* **`print(...)`** — форматированный вывод результатов.

```python
rf = RandomForestClassifier(n_estimators=120, random_state=SEED, n_jobs=-1)

start = time.perf_counter()
rf.fit(X_train, y_train)
rf_train_time = time.perf_counter() - start
rf_pred = rf.predict(X_test)
rf_acc = accuracy_score(y_test, rf_pred)

print(f'Random Forest — Accuracy: {rf_acc:.4f} ({rf_acc:.2%})')
print(f'Время обучения: {rf_train_time:.2f} с')
```

### Задание 2.Построить классификатор SVM и сравнить точность.

В этом блоке мы обучаем метод опорных векторов (SVM) с RBF-ядром, замеряем время обучения, оцениваем качество предсказаний и сравниваем результаты с Random Forest в виде таблицы и графиков.

**Используемые методы и параметры:**
* **`SVC(...)`** — инициализация модели метода опорных векторов из `sklearn.svm`:
  * `kernel='rbf'` — радиальная базисная функция (RBF), нелинейное ядро для построения сложных разделяющих границ
  * `C=5` — параметр регуляризации, контролирует компромисс между шириной зазора и числом ошибок классификации
  * `gamma='scale'` — автоматический подбор ширины ядра по формуле `1 / (n_features × var(X))`
* **`time.perf_counter()`** — высокоточный системный таймер для замера чистого времени обучения модели в секундах
* **`svm_model.fit(X_train, y_train)`** — метод обучения модели на обучающей выборке (матрица признаков `X_train` и вектор меток `y_train`)
* **`svm_model.predict(X_test)`** — метод генерации предсказаний (классов цифр) для тестовой выборки
* **`accuracy_score(y_test, svm_pred)`** — функция из `sklearn.metrics`, вычисляющая метрику Accuracy (долю верно угаданных классов)


```python
svm_model = SVC(kernel='rbf',  C=5,  gamma='scale')

start = time.perf_counter()
svm_model.fit(X_train, y_train)
svm_train_time = time.perf_counter() - start
svm_pred = svm_model.predict(X_test)
svm_acc = accuracy_score(y_test, svm_pred)
```

В коде ниже создаем таблицу сравнения Random Forest и SVM и строим столбчатые диаграммы для сравнения методов по accuracy и времени обучения.

```python
compare_classic = pd.DataFrame({
    'Модель': ['Random Forest', 'SVM (RBF)'],
    'Accuracy': [rf_acc, svm_acc],
    'Время обучения, с': [ rf_train_time, svm_train_time]
})

print(compare_classic.to_string(index=False, float_format=lambda x: f'{x:.4f}'))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(
    compare_classic['Модель'],
    compare_classic['Accuracy']
)
axes[0].set_title('Точность классификаторов')
axes[0].set_ylabel('Accuracy')
axes[0].set_ylim(0, 1)

axes[1].bar(
    compare_classic['Модель'],
    compare_classic['Время обучения, с']
)
axes[1].set_title('Время обучения')
axes[1].set_ylabel('Секунды')

plt.tight_layout()
plt.show()
```

### 3.Реализовать CNN с Dropout для предотвращения переобучения.

В этом блоке реализуется CNN для распознавания рукописных цифр MNIST. Архитектура включает два свёрточных слоя и два слоя Dropout для предотвращения переобучения. После обучения строится визуализация кривых обучения.

**Модель (`Sequential`):** два свёрточных блока `Conv2D` + `MaxPooling2D`, два слоя `Dropout` (0.25 и 0.5) для регуляризации, полносвязный `Dense(128)` и выход `Dense(10, softmax)`.

**Ключевые методы:**
- `compile(optimizer='adam', loss='sparse_categorical_crossentropy')` — настройка обучения
- `fit(epochs=5, batch_size=128, validation_data=...)` — обучение, результат сохраняется в `history`
- `evaluate(...)` — итоговая точность на тесте
- `predict(...)` + `np.argmax(..., axis=1)` — преобразование вероятностей в классы

**Графики:** через `history.history` строятся кривые Accuracy и Loss для обучающей и проверочной выборок — позволяют оценить сходимость и отсутствие переобучения.

```python
cnn = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(32, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),

    Dropout(0.25),

    Flatten(),

    Dense(128, activation='relu'),
    Dropout(0.5),

    Dense(10, activation='softmax')
])
```

```python
cnn.compile( optimizer='adam',  loss='sparse_categorical_crossentropy', metrics=['accuracy'])

cnn.summary()

start = time.perf_counter()

history = cnn.fit( x_train_cnn, y_train, epochs=5, batch_size=128, validation_data=(x_val_cnn, y_val), verbose=1)

cnn_train_time = time.perf_counter() - start

cnn_test_loss, cnn_acc = cnn.evaluate( x_test_cnn, y_test, verbose=0)

cnn_prob = cnn.predict( x_test_cnn, batch_size=256, verbose=0)

cnn_pred = np.argmax(cnn_prob, axis=1)

print(f'CNN — Accuracy: {cnn_acc:.4f} ({cnn_acc:.2%})')
print(f'CNN — Loss: {cnn_test_loss:.4f}')
print(f'Время обучения CNN: {cnn_train_time:.2f} с')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

epochs = np.arange(1, len(history.history['loss']) + 1)

axes[0].plot( epochs, history.history['accuracy'], marker='o', label='Обучение')

axes[0].plot( epochs, history.history['val_accuracy'], marker='o', label='Проверка')

axes[0].set_title('CNN: изменение Accuracy')
axes[0].set_xlabel('Эпоха')
axes[0].set_ylabel('Accuracy')
axes[0].grid(alpha=0.3)
axes[0].legend()

axes[1].plot(epochs,  history.history['loss'], marker='o',  label='Обучение')

axes[1].plot(epochs,  history.history['val_loss'],  marker='o',  label='Проверка')

axes[1].set_title('CNN: изменение Loss')
axes[1].set_xlabel('Эпоха')
axes[1].set_ylabel('Loss')
axes[1].grid(alpha=0.3)
axes[1].legend()

plt.tight_layout()
plt.show()

```

### 4.Проверить влияние размера обучающей выборки на точность модели.

В этом задании мы проводим эксперимент, как качество модели зависит от количества обучающих данных.

**Использованные методы:**
- Обучаем CNN (архитектура из задания 3) на выборках разного размера: 1000, 3000, 5000, 7000, 10000, 30000 изображений
- Для каждого размера используем **стратифицированную выборку** (`stratify=y_train_all`) — сохраняем пропорции классов
- Замеряем время обучения и точность на тестовой выборке с помощью ``time.perf_counter()``
- Строим графики зависимости accuracy и времени обучения от размера выборки

```python
sample_sizes = [1000, 3000, 5000, 7000, 10000, 30000]
accuracies_by_size = []
train_times_by_size = []

for size in sample_sizes:
    print(f"\nОбучение на {size} изображениях...")
    indices, _ = train_test_split(
        np.arange(len(y_train_all)),
        train_size=size,
        random_state=SEED,
        stratify=y_train_all
    )

    x_sub = x_train_all[indices].astype(np.float32) / 255.0
    y_sub = y_train_all[indices]
    x_sub_cnn = x_sub[..., np.newaxis]

    model_temp = Sequential([
        Input(shape=(28, 28, 1)),
        Conv2D(32, (3, 3), activation='relu', padding='same'),
        MaxPooling2D((2, 2)),
        Conv2D(64, (3, 3), activation='relu', padding='same'),
        MaxPooling2D((2, 2)),
        Dropout(0.25),
        Flatten(),
        Dense(128, activation='relu'),
        Dropout(0.5),
        Dense(10, activation='softmax')
    ])

    model_temp.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    # Обучение
    start = time.perf_counter()
    model_temp.fit(
        x_sub_cnn, y_sub,
        epochs=5,
        batch_size=128,
        validation_data=(x_val_cnn, y_val),
        verbose=0
    )
    train_time = time.perf_counter() - start

    # Оценка на тестовой выборке
    _, acc = model_temp.evaluate(x_test_cnn, y_test, verbose=0)
    accuracies_by_size.append(acc)
    train_times_by_size.append(train_time)

    print(f"  Точность: {acc:.4f} ({acc:.2%})")
    print(f"  Время обучения: {train_time:.2f} с")

# Таблица результатов
df_sizes = pd.DataFrame({
    'Размер выборки': sample_sizes,
    'Accuracy': accuracies_by_size,
    'Время обучения, с': train_times_by_size
})
print("\nРезультаты эксперимента:")
print(df_sizes.to_string(index=False, float_format=lambda x: f'{x:.4f}'))

# График зависимости точности от размера выборки
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(sample_sizes, accuracies_by_size, marker='o', linewidth=2,
             markersize=8, color='green')
axes[0].set_title('Зависимость точности от размера выборки')
axes[0].set_xlabel('Размер обучающей выборки')
axes[0].set_ylabel('Accuracy')
axes[0].grid(alpha=0.3)
axes[0].set_xticks(sample_sizes)
for size, acc in zip(sample_sizes, accuracies_by_size):
    axes[0].annotate(f'{acc:.3f}', (size, acc), textcoords="offset points",
                     xytext=(0, 10), ha='center')

axes[1].plot(sample_sizes, train_times_by_size, marker='s', linewidth=2,
             markersize=8, color='orange')
axes[1].set_title('Зависимость времени обучения от размера выборки')
axes[1].set_xlabel('Размер обучающей выборки')
axes[1].set_ylabel('Время, с')
axes[1].grid(alpha=0.3)
axes[1].set_xticks(sample_sizes)

plt.tight_layout()
plt.show()
```

### 5.Сравнить Precision, Recall и F1-score для разных классов цифр.

Для итоговой оценки переобучаем CNN на полном обучающем наборе (10 000 изображений) и вычисляем три ключевые метрики для каждого класса:

- **Precision (точность)** — доля истинно положительных среди всех предсказанных как «класс i»: `TP / (TP + FP)`. Показывает, насколько модель «уверена» в своих предсказаниях.
- **Recall (полнота)** — доля истинно положительных среди всех реально принадлежащих классу i: `TP / (TP + FN)`. Показывает, насколько модель «находит» все примеры класса.
- **F1-score** — гармоническое среднее Precision и Recall, сбалансированная метрика.

**Использованные методы и параметры:**
- `Sequential([...])` + `compile(optimizer='adam', loss='sparse_categorical_crossentropy')` — финальная CNN с той же архитектурой, что и в задании 3.
- `fit(epochs=5, batch_size=128, validation_data=(x_val_cnn, y_val))` — обучение на полном train (10 000 изображений).
- `predict(...)` + `np.argmax(..., axis=1)` — получение вероятностей и перевод их в классы.
- `classification_report(y_test, cnn_final_pred, target_names=[...], digits=4)` — таблица Precision, Recall, F1-score для каждого из 10 классов.
- `ConfusionMatrixDisplay.from_predictions(..., cmap='Blues')` — матрица ошибок с цветовой шкалой.
- `evaluate(...)` — итоговый accuracy и loss на тесте.
- `pd.DataFrame({...})` + `to_string(float_format=...)` — сводная таблица RF / SVM / CNN.
- `plt.subplots(1, 2)` + `bar(...)` + `text(...)` — два столбчатых графика с подписями значений.

```python
print("\nПереобучение CNN на полном датасете для финальных метрик...")
cnn_final = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    Dropout(0.25),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

cnn_final.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

cnn_final.fit(
    x_train_cnn, y_train,
    epochs=5,
    batch_size=128,
    validation_data=(x_val_cnn, y_val),
    verbose=1
)

cnn_final_prob = cnn_final.predict(x_test_cnn, batch_size=256, verbose=0)
cnn_final_pred = np.argmax(cnn_final_prob, axis=1)

print("Отчёт о классификации (Precision / Recall / F1-score):")
report = classification_report(
    y_test, cnn_final_pred,
    target_names=[f'Цифра {i}' for i in range(10)],
    digits=4
)
print(report)

fig, ax = plt.subplots(figsize=(10, 8))
disp = ConfusionMatrixDisplay.from_predictions(
    y_test, cnn_final_pred,
    cmap='Blues',
    ax=ax,
    colorbar=True
)
ax.set_title('Confusion Matrix (CNN с Dropout)', fontsize=14, pad=15)
ax.set_xlabel('Предсказанный класс', fontsize=12)
ax.set_ylabel('Истинный класс', fontsize=12)
plt.tight_layout()
plt.show()

print("Итоговая таблица сравнения всех моделей:")

_, cnn_final_acc = cnn_final.evaluate(x_test_cnn, y_test, verbose=0)

compare_all = pd.DataFrame({
    'Модель': ['Random Forest', 'SVM (RBF)', 'CNN с Dropout'],
    'Accuracy': [rf_acc, svm_acc, cnn_final_acc],
    'Время обучения, с': [rf_train_time, svm_train_time, cnn_train_time]
})
print(compare_all.to_string(index=False, float_format=lambda x: f'{x:.4f}'))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].bar(compare_all['Модель'], compare_all['Accuracy'], color=['#4C72B0', '#DD8452', '#55A868'])
axes[0].set_title('Сравнение точности моделей')
axes[0].set_ylabel('Accuracy')
axes[0].set_ylim(0.90, 1.0)
for i, v in enumerate(compare_all['Accuracy']):
    axes[0].text(i, v + 0.002, f'{v:.4f}', ha='center', fontweight='bold')

axes[1].bar(compare_all['Модель'], compare_all['Время обучения, с'], color=['#4C72B0', '#DD8452', '#55A868'])
axes[1].set_title('Сравнение времени обучения')
axes[1].set_ylabel('Время, с')
for i, v in enumerate(compare_all['Время обучения, с']):
    axes[1].text(i, v + 0.1, f'{v:.2f}', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

### Дополнительный эксперимент: визуализация предсказаний CNN

Для наглядной оценки качества модели отображаем **10 тестовых изображений** с подписями истинного и предсказанного классов.

**Используемые методы:**
- `cnn_final.predict(test_images)` — получение вероятностей по 10 классам для каждого изображения
- `np.argmax(..., axis=1)` — перевод вектора вероятностей в номер класса (максимальная вероятность)
- `plt.subplots(2, 5)` — сетка 2×5 для отображения 10 изображений
- `ax.imshow(..., cmap='gray')` — вывод чёрно-белого изображения
- Цветовое кодирование: **зелёный** — верное предсказание, **красный** — ошибка

**Цель эксперимента:** визуально оценить, на каких типах изображений модель ошибается (например, плохо написанные цифры, похожие пары 4/9, 3/8).

```python
# Берём первые 10 изображений из тестовой выборки
n_samples = 10
test_images = x_test_cnn[:n_samples]
true_labels = y_test[:n_samples]

# Получаем предсказания модели
predictions_prob = cnn_final.predict(test_images, verbose=0)
predicted_labels = np.argmax(predictions_prob, axis=1)

# Визуализация
fig, axes = plt.subplots(2, 5, figsize=(14, 8))

for i, ax in enumerate(axes.ravel()):
    ax.imshow(test_images[i, ..., 0], cmap='gray')
    color = 'green' if predicted_labels[i] == true_labels[i] else 'red'
    ax.set_title(
        f'Истинный: {true_labels[i]}\nПредсказано: {predicted_labels[i]}',
        color=color,
        fontsize=11,
        fontweight='bold'
    )
    ax.axis('off')

plt.suptitle('Предсказания CNN на 10 тестовых изображениях', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

# Статистика по выборке
correct = np.sum(predicted_labels == true_labels)
print(f"\nВерных предсказаний: {correct} из {n_samples} ({correct/n_samples:.0%})")
```

### Ответы на контрольные вопросы

**1. Что такое задача классификации изображений?**

Это задача машинного обучения, цель которой — отнести входное изображение к одному из заранее заданных классов (например, определить, какая цифра от 0 до 9 изображена на картинке).

**2. В чем разница между kNN, SVM и Random Forest для классификации?**

- kNN: Классифицирует объект по «большинству голосов» его k ближайших соседей в пространстве признаков.
- SVM: Ищет оптимальную разделяющую границу (гиперплоскость) между классами. Может использовать «ядра» для работы с нелинейными данными.
- Random Forest: Ансамблевый метод, который строит множество решающих деревьев и усредняет их прогнозы, что снижает риск переобучения.

**3. Какие преимущества у сверточных нейронных сетей (CNN)?**

- Автоматическое извлечение признаков: не нужно создавать их вручную.
- Учет пространственной структуры: сеть понимает, что соседние пиксели связаны (в отличие от обычных полносвязных сетей).
- Инвариантность: устойчивость к небольшим сдвигам, поворотам и искажениям объекта.
- Эффективность: использование общих весов (фильтров) резко сокращает количество обучаемых параметров.

**4. Что такое матрица ошибок (Confusion Matrix)?**

Это таблица, которая показывает количество верных и ошибочных предсказаний модели в разрезе каждого класса. Она позволяет увидеть не просто общую точность, а то, какие именно классы модель путает между собой (например, цифру 4 с цифрой 9).

**5. Какие метрики классификации используются кроме Accuracy?**

- Precision (Точность): Какая доля объектов, предсказанных как «класс А», действительно является «классом А».
- Recall (Полнота): Какая доля реальных объектов «класса А» была успешно найдена моделью.
- F1-score: Гармоническое среднее между Precision и Recall. Это сбалансированная метрика, которая особенно важна при несбалансированных наборах данных (когда одних классов намного больше, чем других).
